```python
import matplotlib.pyplot as plt
import numpy as np
```

# Basic Line Plot

```python
x=np.array([2020,2021,2022,2023,2024])
y=np.array([100,150,110,250,300])

plt.plot(x,y)

plt.title("Line Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

# Line Styles and Colors

```python
x=np.array([1,2,3,4])
y=np.array([5,3,7,8])

plt.plot(y,linestyle="dotted", color="green", marker="o")
plt.title("Styled Line Plot")
plt.savefig("plot2.png")
plt.show()
```

# Multiple Lines in one Graph with Numpy and Matplotlib

```python
x=np.array([1,2,3,4])
y1=np.array([5,3,7,8])
y2=np.array([2,4,1,5])

plt.plot(x,y1,linestyle="dotted", color="green", marker="o")
plt.plot(x,y2,linestyle="--", color="red", marker="s")
plt.title("Multiple Line exapmle")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.legend(["line1","line2"])
plt.show()
```

# Bar Char with Numpy & Matplotib

```python
x=np.array(["A","B","C","D"])
y=np.array([5,3,7,8])

plt.title("Bar chart")
plt.bar(x,y, color="yellowgreen")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

# Scatter plot with Numpy

```python
x=np.array([3,4,5,2])
y=np.array([5,3,7,8])

plt.title("Scatter")
plt.scatter(x,y, color="blue", s=100)
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

# Histogram

```python
data=np.array([3,4,5,2,6,7,8, 19, 14])

plt.title("Histogram")
plt.hist(data, color="blue", bins=8,edgecolor="red")
plt.xlabel("Values")
plt.ylabel("Count")
plt.show()
```

# Subplots

```python
x=np.array([1,2,3,4])
y1=np.array([5,3,7,8])
y2=np.array([2,4,1,5])

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.plot(x,y1, color="green", marker="o")
plt.title("plot 1")

plt.subplot(1,2,2)
plt.bar(x,y2, color="red")
plt.title("plot 2")

plt.show()
```

```python
plt.style.use('_mpl-gallery')

# make data:
np.random.seed(10)
D = np.random.normal((3, 5, 4), (1.25, 1.00, 1.25), (100, 3))

# plot
fig, ax = plt.subplots()
VP = ax.boxplot(D, positions=[2, 4, 6], widths=1.5, patch_artist=True,
                showmeans=False, showfliers=False,
                medianprops={"color": "white", "linewidth": 0.5},

                boxprops={"facecolor": "C0", "edgecolor": "white",
                          "linewidth": 0.5},
                whiskerprops={"color": "C0", "linewidth": 1.5},
                capprops={"color": "C0", "linewidth": 1.5})

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```

```python
plt.style.use('_mpl-gallery-nogrid')


# make data
x = [1, 2, 3, 4]
colors = plt.get_cmap('Blues')(np.linspace(0.2, 0.7, len(x)))

# plot
fig, ax = plt.subplots()
ax.pie(x, colors=colors, radius=3, center=(4, 4),
       wedgeprops={"linewidth": 1, "edgecolor": "white"}, frame=True)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```

```python
plt.style.use('_mpl-gallery')

# Make data
n = 100
xs = np.linspace(0, 1, n)
ys = np.sin(xs * 6 * np.pi)
zs = np.cos(xs * 6 * np.pi)

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot(xs, ys, zs)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```

```python
plt.style.use('_mpl-gallery')

# Make data
n = 20
x = np.sin(np.linspace(0, 2*np.pi, n))
y = np.cos(np.linspace(0, 2*np.pi, n))
z = np.linspace(0, 1, n)

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.stem(x, y, z)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
